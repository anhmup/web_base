```python
import torch
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")  # set device to gpu or cpu
device
```

```python
def split_processing(image):
    image_test = np.array(image).astype(np.uint8)
    h_original , w_original, c = np.array(image_test).shape
    if (h_original > 512):
        threshold_heatmap_row = 0.5
    else :
        threshold_heatmap_row = 0.2
    if (w_original > 400):
        threshold_heatmap_col = 0.3
    else:
        threshold_heatmap_col = 0.2

    if h_original < 100:
        threshold_contour_row = 10
    elif h_original > 512:
        threshold_contour_row = 5
    else:
        threshold_contour_row = 10
    if w_original> 100:
        threshold_contour_col = 10
    else :
        threshold_contour_col = 10
    image = np.array(image_test)
    image_input = cv2.resize(image, (W_SHAPE, H_SHAPE))
    image_input = image_input.astype('float')
    image_input = np.array(image_input,dtype= np.float32)/255.0
    image_tensor =  torch.from_numpy(image_input)
    image_tensor = image_tensor.permute(2,0,1)
    image_tensor = torch.unsqueeze(image_tensor, dim = 0)
    image_tensor.shape
    a,b = split_task(image_tensor.to(device))
    row_list_ = split_model_lib.get_row_pred_line(a , threshold_heatmap_row, threshold_contour_row)
    row_list_ = sorted(row_list_)
    row_list_.insert(0, 1)
    row_list_.append(H_SHAPE-1)
    col_list_ = split_model_lib.get_col_pred_line(b, threshold_heatmap_col, threshold_contour_col)
    # print(row_list_)
    col_list_ = sorted(col_list_)
    col_list_.insert(0, 1)
    col_list_.append(W_SHAPE-1)  # add last column
    split_image = split_model_lib.show_line_pred(row_list_, col_list_, image)
    return split_image , row_list_, col_list_

def merge_processing(image , row_list_, col_list_):
    image_test = np.array(image).astype(np.uint8)  # convert to uint8 type
    h_original, w_original, c = np.array(image_test).shape
    image = np.array(image_test)
    image_input = cv2.resize(image, (W_SHAPE, H_SHAPE))
    image_input = image_input.astype('float')
    image_input = np.array(image_input,dtype= np.float32)/255.0
    image_tensor =  torch.from_numpy(image_input)
    image_tensor = image_tensor.permute(2,0,1)
    image_tensor = torch.unsqueeze(image_tensor, dim = 0)

    cell_bbox_ = merge_model_lib.get_pred_bbox_list(row_list_ , col_list_)
    bbox_input =  torch.from_numpy(np.array([cell_bbox_]))
    row_pred, col_pred = merge_task(image_tensor.to(device), bbox_input.to(device))
    last_cell = bbox_input.numpy()[-1]
    num_row, num_col =  last_cell[-1][0] , last_cell[-1][1]
    row_merge , col_merge  = merge_model_lib.get_final_pred_merge(row_pred, col_pred, num_row, num_col)
    table_dict  = merge_model_lib.get_table(row_merge, col_merge, np.array(cell_bbox_) )

    image_final = merge_model_lib.show_final_result_v2(image , table_dict , h_original, w_original)
    return image_final, table_dict , row_merge , col_merge
```

```python
def is_integer(var):
  return isinstance(var, int)
def get_table_dict_high_res(table_dict_768_512, h_table_high_res, w_table_high_res):
  table_dict_high_res =[]
  for cell in table_dict_768_512:
    cell_dict = dict()
    bbox = cell['bbox']
    x1_768_512 , y1_768_512 , x2_768_512, y2_768_512 = bbox
    x1_high_res = int(x1_768_512/512 * w_table_high_res)
    y1_high_res = int(y1_768_512/768 * h_table_high_res)
    x2_high_res = int(x2_768_512/512 * w_table_high_res)
    y2_high_res = int(y2_768_512/768 * h_table_high_res)
    cell_dict['row'] = cell['row']
    cell_dict['col'] = cell['col']
    cell_dict['bbox'] = [x1_high_res, y1_high_res, x2_high_res, y2_high_res]
    cell_dict['content'] = cell['content']
    table_dict_high_res.append(cell_dict)
  return table_dict_high_res

def show_result(table_image, table_dict):
  table_image_draw = table_image.copy()
  for cell in table_dict:
    bbox = cell['bbox']
    x1, y1, x2, y2 = bbox
    cv2.rectangle(table_image_draw, (x1, y1), (x2, y2), (255, 0, 0), 2)
  return table_image_draw
def get_table_dict_content_v2(table_image, table_dict):
  final_table_dict_2 =[]
  table_image_np = np.array(table_image)
  h_original , w_original, c = np.array(table_image).shape
  for cell in table_dict:
      cell_dict = dict()
      bbox = cell['bbox']
      x1, y1, x2, y2 = bbox
      row_cnt = cell['row']
      col_cnt = cell['col']
      content = ''
      cell_dict['row'] = row_cnt
      cell_dict['col'] = col_cnt
      # print(f' num :{row_cnt},  {col_cnt}')
      # cell_dict['bbox'] = [x1, y1, x2, y2]
      xmin = x1 -2
      if xmin < 0 :
        xmin = 0
      xmax = x2 + 2
      if xmax > w_original:
        xmax = w_original
      ymin = y1 -2
      if ymin < 0 :
        ymin = 0
      ymax = y2 + 2
      if ymax > h_original:
        ymax = h_original
      sub_image = table_image_np[ymin:ymax, xmin:xmax]

      scale_factor = 4  # Adjust as needed
      resized_image = cv2.resize(sub_image, (sub_image.shape[1] * scale_factor, sub_image.shape[0] * scale_factor), interpolation=cv2.INTER_CUBIC)
      text = pytesseract.image_to_string(resized_image, lang='vie')
      try:
        content = str(text)
      except:
        content = ' '
      cell_dict['content'] = content
      final_table_dict_2.append(cell_dict)
  return final_table_dict_2

def get_html_from_wb(final_table_dict_2, num_row, num_col):
  row_max = num_row
  col_max = num_col
  # print(row_max, col_max)
  wb = openpyxl.Workbook()
  sheet = wb.active
  for data_cell in final_table_dict_2:
    row_cnt = data_cell['row']
    col_cnt = data_cell['col']
    try:
      start_row = int(data_cell['row']) + 1
    except:
      start_row = int(data_cell['row'][0]) + 1
    try:
      start_col = int(data_cell['col']) + 1
    except:
      start_col = int(data_cell['col'][0]) + 1
    content = data_cell['content']
    if not content:
      content = ' '
    content = "".join(c for c in content if c.isprintable()) # keep only printable characters
    sheet.cell(row=start_row, column=start_col, value=content)
    sheet.cell(row=start_row, column=start_col).alignment = Alignment(horizontal='center', vertical='center')
  # merge cell
  for data_cell in final_table_dict_2:
    # check spanning cell
    row_merge_flag = 0
    col_merge_flag = 0
    row_cnt = data_cell['row']
    col_cnt = data_cell['col']
    # print(row_cnt)
    # print(col_cnt)
    if not is_integer(row_cnt):
      row_merge_flag = 1
    if not is_integer(col_cnt):
      col_merge_flag = 1
    if col_merge_flag == 1 and row_merge_flag == 0:
      start_col = int(data_cell['col'][0]) + 1
      end_col = int(data_cell['col'][-1]) + 1
      start_row = int(data_cell['row']) + 1
      end_row = int(data_cell['row']) + 1
      sheet.merge_cells(start_row=start_row, start_column=start_col, end_row=end_row, end_column=end_col)
    if col_merge_flag == 0 and row_merge_flag == 1:
      start_col = int(data_cell['col']) + 1
      end_col = int(data_cell['col']) + 1
      start_row = int(data_cell['row'][0]) + 1
      end_row = int(data_cell['row'][-1]) + 1
      sheet.merge_cells(start_row=start_row, start_column=start_col, end_row=end_row, end_column=end_col)
    if col_merge_flag == 1 and row_merge_flag == 1:
      start_col = int(data_cell['col'][0]) + 1
      end_col = int(data_cell['col'][-1]) + 1
      start_row = int(data_cell['row'][0]) + 1
      end_row = int(data_cell['row'][-1]) + 1
      sheet.merge_cells(start_row=start_row, start_column=start_col, end_row=end_row, end_column=end_col)
  # wb.save('test.xlsx')
  html_final = workbook_to_html(wb)
  return html_final


def workbook_to_html(wb, sheet_name=0):
    ws = wb[wb.sheetnames[sheet_name]] if isinstance(sheet_name, int) else wb[sheet_name]
    merged_cells = ws.merged_cells.ranges  # Get merged cell ranges

    html = '<table border="1">\n'
    for i, row in enumerate(ws.iter_rows(values_only=True)):
        html += "  <tr>\n"
        for j, cell in enumerate(row):
            rowspan, colspan = 1, 1
            for merged in merged_cells:
                if ws.cell(row=i+1, column=j+1).coordinate in merged:
                    rowspan = merged.size["rows"]
                    colspan = merged.size["columns"]
                    if i+1 != merged.min_row or j+1 != merged.min_col:
                        cell = None  # Skip duplicate merged cells

            if cell is not None:
                attributes = ""
                if rowspan > 1:
                    attributes += f' rowspan="{rowspan}"'
                if colspan > 1:
                    attributes += f' colspan="{colspan}"'
                html += f'    <td{attributes}>{cell}</td>\n'

        html += "  </tr>\n"
    html += "</table>"

    return html
```
